## 1) Descripción del problema **[1 pts]**


Una tienda necesita un sistema que permita administrar la atencion de sus clientes y su catalogo de productos.

Los clientes deben ser atendidos en el mismo oden en el que llegan, por lo que se utilizara una cola. Cada cliente que sea atendido se guardara en una pila, permitiendo consultar o deshacer la ultima atencion realizada.

Ademas, la tienda necesita administrar sus productos mediante una lista doblemente enlazada, lo que permitira recorrer el catalogo desde el primer producto hasta el ultimo y tambien en sentido contrario, ademas de buscar y eliminar productos mediante su codigo.

```python



print("==========================================")
# ESTRUCTURAS DE DATOS BASE
print("==========================================")


class NodoCliente:
    def __init__(self, nombre):
        self.nombre = nombre
        self.siguiente = None


class ColaClientes:
    def __init__(self):
        self.frente = None
        self.final = None

    def esta_vacia(self):
        return self.frente is None

    def encolar(self, nombre):
        nuevo_nodo = NodoCliente(nombre)
        if self.esta_vacia():
            self.frente = nuevo_nodo
            self.final = nuevo_nodo
        else:
            self.final.siguiente = nuevo_nodo
            self.final = nuevo_nodo

    def encolar_al_inicio(self, nombre):
        nuevo_nodo = NodoCliente(nombre)
        if self.esta_vacia():
            self.frente = nuevo_nodo
            self.final = nuevo_nodo
        else:
            nuevo_nodo.siguiente = self.frente
            self.frente = nuevo_nodo

    def desencolar(self):
        if self.esta_vacia():
            return None
        nombre = self.frente.nombre
        self.frente = self.frente.siguiente
        if self.frente is None:
            self.final = None
        return nombre

    def mostrar(self):
        if self.esta_vacia():
            print("  [ Fila vacía ]")
            return
        actual = self.frente
        posicion = 1
        print("  Fila de espera actual:")
        while actual:
            print(f"    {posicion}. {actual.nombre}")
            actual = actual.siguiente
            posicion += 1


class NodoPila:
    def __init__(self, cliente):
        self.cliente = cliente
        self.siguiente = None


class PilaHistorial:
    def __init__(self):
        self.tope = None

    def esta_vacia(self):
        return self.tope is None

    def apilar(self, cliente):
        nuevo_nodo = NodoPila(cliente)
        nuevo_nodo.siguiente = self.tope
        self.tope = nuevo_nodo

    def desapilar(self):
        if self.esta_vacia():
            return None
        cliente = self.tope.cliente
        self.tope = self.tope.siguiente
        return cliente

    def ver_ultimo(self):
        if self.esta_vacia():
            return None
        return self.tope.cliente


class NodoProducto:
    def __init__(self, codigo, nombre, precio):
        self.codigo = codigo
        self.nombre = nombre
        self.precio = precio
        self.siguiente = None
        self.anterior = None


class ListaDobleProductos:
    def __init__(self):
        self.cabeza = None
        self.cola = None

    def existe_codigo(self, codigo):
        actual = self.cabeza
        while actual:
            if actual.codigo == codigo:
                return True
            actual = actual.siguiente
        return False

    def agregar(self, codigo, nombre, precio):
        if self.existe_codigo(codigo):
            print(f"Error: Ya existe un producto registrado con el código {codigo}.")
            return False

        nuevo_nodo = NodoProducto(codigo, nombre, precio)
        if self.cabeza is None:
            self.cabeza = nuevo_nodo
            self.cola = nuevo_nodo
        else:
            self.cola.siguiente = nuevo_nodo
            nuevo_nodo.anterior = self.cola
            self.cola = nuevo_nodo
        print(f"Producto '{nombre}' agregado exitosamente con el código '{codigo}'.")
        return True

    def mostrar_desde_inicio(self):
        if self.cabeza is None:
            print("  [ Catálogo de productos vacío ]")
            return
        actual = self.cabeza
        print("  Catálogo de productos (Del primero al último):")
        while actual:
            print(f"    Código: {actual.codigo} | Nombre: {actual.nombre} | Precio: ${actual.precio:.2f}")
            actual = actual.siguiente

    def mostrar_desde_final(self):
        if self.cola is None:
            print("  [ Catálogo de productos vacío ]")
            return
        actual = self.cola
        print("  Catálogo de productos (Del último al primero):")
        while actual:
            print(f"    Código: {actual.codigo} | Nombre: {actual.nombre} | Precio: ${actual.precio:.2f}")
            actual = actual.anterior

    def buscar_por_codigo(self, codigo):
        actual = self.cabeza
        while actual:
            if actual.codigo == codigo:
                return actual
            actual = actual.siguiente
        return None

    def eliminar_por_codigo(self, codigo):
        actual = self.buscar_por_codigo(codigo)
        if actual is None:
            print(f"Error: No se encontró ningún producto con el código '{codigo}'.")
            return False

        if actual.anterior:
            actual.anterior.siguiente = actual.siguiente
        else:
            self.cabeza = actual.siguiente

        if actual.siguiente:
            actual.siguiente.anterior = actual.anterior
        else:
            self.cola = actual.anterior

        print(f"Producto '{actual.nombre}' con código '{codigo}' eliminado exitosamente del catálogo.")
        return True


print("==========================================")
# SISTEMA DE GESTIÓN DE LA TIENDA
print("==========================================")


class SistemaTienda:
    def __init__(self):
        self.fila_espera = ColaClientes()
        self.historial_atencion = PilaHistorial()
        self.catalogo_productos = ListaDobleProductos()

    def registrar_cliente(self):
        nombre = input("Ingrese el nombre del cliente: ").strip()
        if not nombre:
            print("El nombre del cliente no puede estar vacío.")
            return
        self.fila_espera.encolar(nombre)
        print(f"Cliente '{nombre}' agregado exitosamente al final de la fila de espera.")

    def atender_cliente(self):
        if self.fila_espera.esta_vacia():
            print("Aviso: No hay clientes esperando en la fila.")
            return
        cliente = self.fila_espera.desencolar()
        self.historial_atencion.apilar(cliente)
        print(f"Atendiendo al cliente: '{cliente}'. Se ha guardado en el historial de atenciones.")

    def mostrar_fila_espera(self):
        self.fila_espera.mostrar()

    def ver_ultimo_cliente_atendido(self):
        cliente = self.historial_atencion.ver_ultimo()
        if cliente is None:
            print("Aviso: No hay atenciones registradas en el historial.")
        else:
            print(f"El último cliente atendido fue: '{cliente}'.")

    def cancelar_ultima_atencion(self):
        if self.historial_atencion.esta_vacia():
            print("Aviso: No hay ninguna atención registrada para deshacer.")
            return
        cliente = self.historial_atencion.desapilar()
        self.fila_espera.encolar_al_inicio(cliente)
        print(f"Atención cancelada. El cliente '{cliente}' ha regresado al comienzo de la fila de espera.")

    def agregar_producto(self):
        codigo = input("Ingrese el código único del producto: ").strip()
        if not codigo:
            print("El código no puede estar vacío.")
            return
        nombre = input("Ingrese el nombre del producto: ").strip()
        if not nombre:
            print("El nombre no puede estar vacío.")
            return
        try:
            precio = float(input("Ingrese el precio del producto: "))
            if precio < 0:
                print("El precio no puede ser negativo.")
                return
        except ValueError:
            print("Entrada inválida. El precio debe ser un número.")
            return

        self.catalogo_productos.agregar(codigo, nombre, precio)

    def ver_productos_inicio_a_final(self):
        self.catalogo_productos.mostrar_desde_inicio()

    def ver_productos_final_a_inicio(self):
        self.catalogo_productos.mostrar_desde_final()

    def buscar_producto(self):
        codigo = input("Ingrese el código del producto a buscar: ").strip()
        producto = self.catalogo_productos.buscar_por_codigo(codigo)
        if producto:
            print(f"Producto encontrado: Código: {producto.codigo} | Nombre: {producto.nombre} | Precio: ${producto.precio:.2f}")
        else:
            print(f"Aviso: El producto con código '{codigo}' no fue encontrado en el catálogo.")

    def eliminar_producto(self):
        codigo = input("Ingrese el código del producto a eliminar: ").strip()
        self.catalogo_productos.eliminar_por_codigo(codigo)


 print("==========================================")
# MENÚ INTERACTIVO
print("==========================================")


def menu():
    tienda = SistemaTienda()

    while True:
        print("\n" + "=" * 50)
        print("    SISTEMA DE GESTIÓN DE TIENDA Y CATÁLOGO")
        print("=" * 50)
        print("--- GESTIÓN DE CLIENTES ---")
        print("1. Registrar nuevo cliente en la fila")
        print("2. Atender cliente")
        print("3. Mostrar clientes en espera")
        print("4. Ver último cliente atendido")
        print("5. Cancelar última atención realizada")
        print("\n--- GESTIÓN DE PRODUCTOS ---")
        print("6. Agregar producto")
        print("7. Ver productos (Primero -> Último)")
        print("8. Ver productos (Último -> Primero)")
        print("9. Buscar producto por código")
        print("10. Eliminar producto por código")
        print("\n0. Salir")
        print("=" * 50)

        opcion = input("Seleccione una opción: ").strip()

        if opcion == "1":
            tienda.registrar_cliente()
        elif opcion == "2":
            tienda.atender_cliente()
        elif opcion == "3":
            tienda.mostrar_fila_espera()
        elif opcion == "4":
            tienda.ver_ultimo_cliente_atendido()
        elif opcion == "5":
            tienda.cancelar_ultima_atencion()
        elif opcion == "6":
            tienda.agregar_producto()
        elif opcion == "7":
            tienda.ver_productos_inicio_a_final()
        elif opcion == "8":
            tienda.ver_productos_final_a_inicio()
        elif opcion == "9":
            tienda.buscar_producto()
        elif opcion == "10":
            tienda.eliminar_producto()
        elif opcion == "0":
            print("Saliendo del sistema...")
            break
        else:
            print("Opción no válida. Intente nuevamente.")


if __name__ == "__main__":
    menu()
```
