## 1) Descripción del problema **[1 pts]**


El vivero requiere calcular la cantidad de nuevas plantas producidas en un mes $n$. Se define una relación de recurrencia basada en la sucesión de Fibonacci: $T(0) = 0$, $T(1) = 1$ y $T(n) = T(n-1) + T(n-2)$ para $n \ge 2$. El objetivo principal es optimizar este cálculo mediante programación dinámica empleando memoización (top-down) para evitar recálculos innecesarios de subproblemas y garantizar un tiempo de ejecución eficiente.

### 5) Análisis de complejidad **[15 pts]**

Función de complejidad $T(n)$:Sin memoización, la recurrencia es $T(n) = T(n-1) + T(n-2) + O(1)$, con una complejidad exponencial $O(2^n)$,

Con memoización, cada subproblema desde $0$ hasta $n$ se resuelve únicamente una vez.

Al procesar el estado $k$, la llamada $T(k-1)$ calcula los estados faltantes y $T(k-2)$ se resuelve en tiempo constante $O(1)$ por búsqueda en caché.Justificación:Complejidad Temporal: $O(n)$.

Existen $n+1$ estados únicos, y cada uno toma tiempo $O(1)$ para combinarse y almacenarse.Complejidad Espacial: $O(n)$ debido a la profundidad máxima de la pila de llamadas recursivas y al espacio utilizado por la memoria (mapa o arreglo).

Mejor y Peor Caso :Mejor Caso: $O(1)$, cuando $n=0$, $n=1$, o cuando la respuesta para $n$ ya se encuentra en la memoria previamente rellenada.

Peor Caso: $O(n)$, cuando la memoria está vacía y se deben computar linealmente los $n$ niveles hacia arriba.

## 6) Código funcional **[15 pts]**
Incluye tu implementación completa.

No pegues código incompleto; debe ser ejecutable.


```python
def calcular_plantas_vivero(n: int, memo: dict = None) -> int:
    """
    Calcula el número de plantas producidas en el mes n usando memoización.
    Recurrencia: T(n) = T(n-1) + T(n-2)
    """
    if memo is None:
        memo = {0: 0, 1: 1}

    if n in memo:
        return memo[n]

    memo[n] = calcular_plantas_vivero(n - 1, memo) + calcular_plantas_vivero(n - 2, memo)
    return memo[n]

mes_usuario = int(input("Ingrese el número de mes: "))


resultado = calcular_plantas_vivero(mes_usuario)
print(f"En el mes {mes_usuario} se producirán {resultado} nuevas plantas.")

```
