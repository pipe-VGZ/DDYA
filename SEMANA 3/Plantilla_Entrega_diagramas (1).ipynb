## 7) Análisis de complejidad **[15 pts]**

7) Análisis de complejidad

1. Ordenamiento: Merge Sort

Justificación:
Merge Sort funciona dividiendo la lista en partes cada vez más pequeñas hasta que quedan elementos individuales. Después va juntando las partes y comparando los elementos para dejarlos ordenados. Por eso utiliza la técnica de Dividir y Conquistar

Función T(n):
La forma de representar su funcionamiento es:

T(n) = 2T(n/2) + O(n)

Esto significa que la lista se divide en dos partes y luego se necesita tiempo para unirlas y ordenarlas

Mejor y peor caso:
En Merge Sort el comportamiento es prácticamente el mismo sin importar cómo este organizada la lista al principio

- Mejor caso: Θ( log n)
- Peor caso: Θ(n2)



2. Búsqueda: Búsqueda Binaria

Justificación:
La Búsqueda Binaria busca un elemento comparándolo con el que está en la mitad de la lista, dependiendo de si el código buscado es menor o mayor, se descarta una de las dos partes y se continúa buscando en la parte que queda

Función T(n):
Su ecuación de recurrencia es:

T(n) = T(n/2) + O(1)

esto pasa porque en cada paso se reduce la lista aproximadamente a la mitad

Mejor y peor caso:

- Mejor caso: Ω(1), cuando el código que buscamos está justo en la mitad y se encuentra en el primer intento.
- Peor caso: O(log n), cuando el código no está en la lista o se encuentra después de varias divisiones image.png

```python
def merge_sort(lista):

    if len(lista) <= 1:
        return lista


    medio = len(lista) // 2
    mitad_izquierda = lista[:medio]
    mitad_derecha = lista[medio:]


    izquierda_ordenada = merge_sort(mitad_izquierda)
    derecha_ordenada = merge_sort(mitad_derecha)

    return combinar(izquierda_ordenada, derecha_ordenada)


def combinar(izquierda, derecha):

    lista_combinada = []
    i = 0
    j = 0


    while i < len(izquierda) and j < len(derecha):
        if izquierda[i] <= derecha[j]:
            lista_combinada.append(izquierda[i])
            i += 1
        else:
            lista_combinada.append(derecha[j])
            j += 1


    lista_combinada.extend(izquierda[i:])
    lista_combinada.extend(derecha[j:])

    return lista_combinada


def busqueda_binaria(lista, codigo_buscado):

    inicio = 0
    fin = len(lista) - 1

    while inicio <= fin:
        medio = (inicio + fin) // 2

        if lista[medio] == codigo_buscado:
            return True
        elif lista[medio] < codigo_buscado:
            inicio = medio + 1
        else:
            fin = medio - 1

    return False


def programa_principal():
    print(" PROGRAMA DE GESTIÓN Y BÚSQUEDA DE PRODUCTOS \n")

    try:
        n = int(input("Ingrese la cantidad de productos: "))
        if n <= 0:
            print("Debe ingresar un número mayor a cero.")
            return

        codigos_originales = []
        for i in range(1, n + 1):
            codigo = int(input(f"Ingrese el código del producto {i}: "))
            codigos_originales.append(codigo)

        print("\n RESUMEN ")
        print(f"Lista original: {codigos_originales}")


        codigos_ordenados = merge_sort(codigos_originales)
        print(f"Lista ordenada (Merge Sort): {codigos_ordenados}\n")


        codigo_a_buscar = int(input("Ingrese el código que desea buscar: "))
        encontrado = busqueda_binaria(codigos_ordenados, codigo_a_buscar)

        if encontrado:
            print(f"El producto con código {codigo_a_buscar} fue encontrado.")
        else:
            print(f"El producto con código {codigo_a_buscar} no se encuentra registrado.")

    except ValueError:
        print("\n[Error]: Debe ingresar únicamente números enteros válidos.")


if __name__ == "__main__":
    programa_principal()
```
