### 5) Análisis de complejidad **[15 pts]**
El proyecto busca conectar los siete componentes de una protoboard utilizando la menor cantidad de cable posible. Para ello, se representa el problema mediante un grafo ponderado y se utiliza el algoritmo de Kruskal para encontrar el Árbol de Expansión Mínima (MST).

5.1. Variables de complejidad

Para el análisis se utilizan dos variables:

V: cantidad de vértices o componentes.

E: cantidad de aristas o conexiones posibles.

Como todos los componentes pueden conectarse entre sí, se utiliza la fórmula:

E = V * (V - 1) / 2

En este laboratorio existen 7 componentes:

V = 7

E = 7 * 6 / 2 = 21

Por lo tanto, se generan 21 conexiones posibles y se seleccionan únicamente 6 para construir el MST.

5.2. Complejidad de las operaciones

Calcular distancias: O(1). Realiza operaciones matemáticas simples entre dos posiciones.

Construir el grafo: O(V²). Compara todas las parejas posibles de componentes.

Ordenar las aristas: O(E log E). Organiza las conexiones de menor a mayor peso.

Detectar ciclos: O(1) amortizado. Utiliza conjuntos disjuntos para comprobar si dos componentes ya están conectados.

Seleccionar el MST: O(E). Recorre las conexiones hasta encontrar las V - 1 necesarias.

Calcular longitud total: O(V). Suma los pesos de los cables seleccionados.

La operación más costosa es el ordenamiento de las aristas, porque debe organizar todas las conexiones antes de ejecutar la selección.

5.3. Complejidad temporal

La construcción del grafo tiene una complejidad de O(V²), mientras que el ordenamiento tiene una complejidad de O(E log E).

Como el grafo es completo y E depende de V², la complejidad general es:

O(V² log V)

Esta representa el crecimiento del tiempo de ejecución del proyecto.

5.4. Mejor caso

El mejor caso ocurre cuando las conexiones ya están ordenadas de menor a mayor peso. Python puede aprovechar ese orden para realizar el proceso de ordenamiento más rápidamente.

Sin embargo, el programa siempre debe calcular todas las distancias entre los componentes.

Por lo tanto, su complejidad es:

O(V²)

Justificacion: aunque el ordenamiento sea más rápido, la construcción de todas las conexiones sigue requiriendo comparar cada pareja de componentes.

5.5. Peor caso

El peor caso ocurre cuando las conexiones necesitan un ordenamiento completo y Kruskal debe revisar una gran cantidad de aristas antes de completar el árbol.

Su complejidad es:

O(V² log V)

Justificacion: el programa debe construir todas las conexiones, ordenarlas y comprobar cuáles pueden seleccionarse sin formar ciclos.

5.6. Complejidad espacial

La complejidad espacial representa la cantidad de memoria que necesita el programa.

Se almacenan los componentes, sus posiciones, todas las conexiones posibles y las conexiones seleccionadas para el MST.

Por lo tanto, la complejidad espacial es:

O(V²)

Justificacion: la mayor cantidad de memoria se utiliza para almacenar las conexiones posibles entre todos los componentes.

5.7. Resumen del análisis

Mejor caso temporal: O(V²).

Peor caso temporal: O(V² log V).

Complejidad espacial: O(V²).

Conclusión:

El algoritmo de Kruskal es adecuado para este proyecto porque permite conectar todos los componentes utilizando la menor longitud total de cable y evitando ciclos. Su complejidad es manejable para los siete componentes del laboratorio y permite trabajar con grafos de mayor tamaño sin aumentar excesivamente el tiempo de procesamiento.

```python
import matplotlib.pyplot as plt


pos = {
    'F': (3, 1),
    'L': (8, 3),
    'S': (16, 2),
    'P': (24, 4),
    'B': (6, 8),
    'R': (15, 9),
    'M': (25, 7)
}


def calcular_distancia(p1, p2):
    return abs(p1[0] - p2[0]) + abs(p1[1] - p2[1])


def obtener_mst(current_pos):
    nodos = list(current_pos.keys())
    aristas = []


    for i in range(len(nodos)):
        for j in range(i + 1, len(nodos)):
            u = nodos[i]
            v = nodos[j]
            peso = calcular_distancia(current_pos[u], current_pos[v])
            aristas.append((peso, u, v))


    aristas.sort(key=lambda x: x[0])


    padre = {n: n for n in nodos}

    def encontrar(i):
        if padre[i] == i:
            return i
        padre[i] = encontrar(padre[i])
        return padre[i]

    def unir(i, j):
        raiz_i = encontrar(i)
        raiz_j = encontrar(j)
        if raiz_i != raiz_j:
            padre[raiz_i] = raiz_j
            return True
        return False

    mst_aristas = []
    total_len = 0

    for peso, u, v in aristas:
        if unir(u, v):
            mst_aristas.append((u, v, peso))
            total_len += peso
            if len(mst_aristas) == len(nodos) - 1:
                break

    return mst_aristas, total_len, aristas


print("=== TODAS LAS POSIBLES CONEXIONES ===")
_, _, todas_aristas = obtener_mst(pos)
for peso, u, v in todas_aristas:
    print(f"{u} -------- {peso} -------- {v}")


mst_original, len_original, _ = obtener_mst(pos)

print("\n========== RESULTADO MST ORIGINAL ==========")
for idx, (u, v, p) in enumerate(mst_original, 1):
    print(f"Cable {idx}: {u} - {v} = {p}")
print(f"Longitud total de cable: {len_original} unidades")


pos_nuevo = pos.copy()
pos_nuevo['B'] = (20, 10)

mst_nuevo, len_nuevo, _ = obtener_mst(pos_nuevo)

print("\n========== RESULTADO MST DESPUÉS DEL CAMBIO (Buzzer en J20) ==========")
for idx, (u, v, p) in enumerate(mst_nuevo, 1):
    print(f"Cable {idx}: {u} - {v} = {p}")
print(f"Longitud total de cable: {len_nuevo} unidades")

print(f"\n--- COMPARATIVA FINAL ---")
print(f"MST original Longitud total: {len_original}")
print(f"MST después del cambio Longitud total: {len_nuevo}")
if len_nuevo > len_original:
    print("Resultado: La cantidad total de cable AUMENTÓ.")
elif len_nuevo < len_original:
    print("Resultado: La cantidad total de cable DISMINUYÓ.")
else:
    print("Resultado: La cantidad total de cable se MANTUVO IGUAL.")


def dibujar_protoboard(current_pos, mst_edges, titulo):
    fig, ax = plt.subplots(figsize=(14, 6))


    for u, v, p in mst_edges:
        p1 = current_pos[u]
        p2 = current_pos[v]
        ax.plot([p1[0], p2[0]], [p1[1], p2[1]], color='blue', linestyle='-', linewidth=2, zorder=1)

        mid_x = (p1[0] + p2[0]) / 2
        mid_y = (p1[1] + p2[1]) / 2
        ax.text(mid_x, mid_y, str(p), color='red', fontsize=10, fontweight='bold',
                bbox=dict(facecolor='white', alpha=0.7, edgecolor='none'))


    for nodo, (col, fila) in current_pos.items():
        ax.scatter(col, fila, s=500, color='orange', edgecolor='black', zorder=2)
        ax.text(col, fila, nodo, fontsize=12, fontweight='bold', ha='center', va='center', zorder=3)

    ax.set_xlim(0, 31)
    ax.set_ylim(11, 0)
    ax.set_xticks(range(1, 31))
    ax.set_yticks(range(1, 11))
    ax.set_yticklabels(['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H', 'I', 'J'])
    ax.grid(True, linestyle='--', alpha=0.5)
    ax.set_title(titulo, fontsize=14)
    ax.set_xlabel("Columnas (1-30)")
    ax.set_ylabel("Filas (A-J)")
    plt.show()

```
